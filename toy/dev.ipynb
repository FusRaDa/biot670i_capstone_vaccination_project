### Let's now look at toy data on restaurant tips

````python
import seaborn as sns

# Load the dataset
df = sns.load_dataset('tips')

# View the first few rows
df.head()
````

````python
import plotly.express as px

day = df['day']
tip = df['tip']
sex = df['sex']


day_tip_sex_gr = px.histogram(df, x=day, y=tip,
             color=sex, barmode='group',
             color_discrete_map={"Male": "blue", "Female": "red"},
             height=400)
day_tip_sex_gr.show()
````

### What days make the most revenue?

````python
day = df['day']
total_bill = df['total_bill']
time = df['time']


day_total_time_gr = px.histogram(df, x=day, y=total_bill,
             color=time, barmode='group',
             height=400)
day_total_time_gr.show()
````
