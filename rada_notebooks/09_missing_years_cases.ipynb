```python
import requests
import pdfplumber
from io import BytesIO

year = 2011
pdf_url = pdf_urls[year]
pdf_url

response = requests.get(pdf_url)
response.raise_for_status()

pdf_bytes = BytesIO(response.content)

with pdfplumber.open(pdf_bytes) as pdf:

    print("Pages:", len(pdf.pages))

    for page_num, page in enumerate(pdf.pages, start=1):

        text = page.extract_text() or ""

        if "measles" in text.lower() or "mumps" in text.lower():

            print(f"\n{'=' * 60}")
            print(f"PAGE {page_num}")
            print("=" * 60)

            # Show lines containing our diseases
            for line in text.splitlines():
                if "measles" in line.lower() or "mumps" in line.lower():
                    print(line)
```

```python
import pandas as pd
import pdfplumber
from io import BytesIO

# pd.set_option("display.max_rows", None)

page_numbers = {
    "measles": [51, 52],
    "mumps": [53, 54]
}

table_settings = {
    "vertical_strategy": "text",
    "horizontal_strategy": "text",
    "snap_tolerance": 3,
    "join_tolerance": 3,
    "intersection_tolerance": 3,
}

with pdfplumber.open(BytesIO(response.content)) as pdf:

    for disease, pages in page_numbers.items():

        for i, page_number in enumerate(pages, start=1):

            # Account for 2-page PDF offset
            pdf_page_number = page_number + 2

            page = pdf.pages[pdf_page_number - 1]

            tables = page.extract_tables(table_settings)

            print(f"\n{'=' * 50}")
            print(f"{disease.upper()} - PRINTED PAGE {page_number}")
            print(f"{'=' * 50}")
            print("Tables found:", len(tables))

            df = pd.DataFrame(tables[0])

            if disease == "measles":
                if i == 1:
                    df_measles_1 = df.copy()
                else:
                    df_measles_2 = df.copy()

            elif disease == "mumps":
                if i == 1:
                    df_mumps_1 = df.copy()
                else:
                    df_mumps_2 = df.copy()

            display(df.head(10))
```

```python
regions = [
    "United States",
    "New England",
    "Mid. Atlantic",
    "E.N. Central",
    "W.N. Central",
    "S. Atlantic",
    "E.S. Central",
    "W.S. Central",
    "Mountain",
    "Pacific"
]

territories = [
    "American Samoa",
    "C.N.M.I.",
    "Guam",
    "Puerto Rico",
    "U.S. Virgin Islands"
]

case_df = case_df[
    ~case_df["state"].isin(regions + territories)
].copy()
```

```python
import numpy as np
import pandas as pd

case_cols = [
    "measles_indigenous",
    "measles_imported",
    "mumps_cases"
]

for col in case_cols:

    # Clean CDC symbols
    case_df[col] = (
        case_df[col]
        .astype(str)
        .str.strip()
        .replace({
            "-": "0",       # No reported cases
            "—": "0",
            "N": np.nan,    # Not notifiable
            "U": np.nan     # Unavailable
        })
    )

    # IMPORTANT: convert strings to numbers
    case_df[col] = pd.to_numeric(
        case_df[col],
        errors="coerce"
    )
```

```python
case_df["measles_cases"] = (
    case_df["measles_indigenous"]
    + case_df["measles_imported"]
)
```

```python
ny_rows = case_df[
    case_df["state"].isin([
        "New York (Upstate)",
        "New York City"
    ])
]

ny = pd.DataFrame({
    "state": ["N.Y."],
    "measles_indigenous": [ny_rows["measles_indigenous"].sum()],
    "measles_imported": [ny_rows["measles_imported"].sum()],
    "mumps_cases": [ny_rows["mumps_cases"].sum()],
    "measles_cases": [ny_rows["measles_cases"].sum()]
})

case_df = case_df[
    ~case_df["state"].isin([
        "New York (Upstate)",
        "New York City"
    ])
]

case_df = pd.concat(
    [case_df, ny],
    ignore_index=True
)
```
