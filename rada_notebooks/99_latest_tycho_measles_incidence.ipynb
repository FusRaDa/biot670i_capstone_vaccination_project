# Clean data for latest measles incidience 

Donwload data here: https://zenodo.org/records/11452259?utm_source=chatgpt.com

```python
import pandas as pd

core_columns = [
    'Admin1ISO',
    'PeriodEndDate',
    'PartOfCumulativeCountSeries',
    'CountValue'
]

measles_df = pd.read_csv('../raw/tycho/US.14189004.csv')
measles_df = measles_df[core_columns]
measles_df.head()
```

```python
len(measles_df)
```

```python
measles_df = measles_df[measles_df['PartOfCumulativeCountSeries'] == 0]
measles_df['year'] = measles_df['PeriodEndDate'].str.split('-').str[0].astype(int)
measles_df['state'] = measles_df['Admin1ISO'].str.split('-').str[1]
measles_df.head()
```

```python
measles_df.drop(columns=['Admin1ISO', 'PeriodEndDate', 'PartOfCumulativeCountSeries'], inplace=True)
measles_df.rename(columns={'CountValue': 'cases'}, inplace=True)
measles_df.head()
```

```python
measles_df.to_csv('../app/data/latest_measles_cases.csv', index=False)
```
