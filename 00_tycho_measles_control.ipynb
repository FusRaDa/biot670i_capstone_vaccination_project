# First step of this project is to reproduce Project Tycho's measles report from 1963.
- Dataset: https://www.tycho.pitt.edu/dataset/US.14189004/?utm_source=chatgpt.com
- Article: https://academic.oup.com/ofid/article/5/7/ofy137/5039595

```python
import os
from dotenv import load_dotenv

# Load variables from .env file into the environment
load_dotenv()

# Access the variables
API_KEY = os.getenv("TYCHO_API_KEY")
```

### Columnns needed from this query:
- ConditionName
- ConditionSNOMED
- CountryCode
- Admin1ISO
- Admin1Name
- PeriodStartDate
- PeriodEndDate
- CountValue
- PartOfCumulativeCountSeries
- Fatalities

```python
core_columns = [
    "ConditionName",
    "ConditionSNOMED",
    "CountryCode",
    "Admin1ISO",
    "Admin1Name",
    "PeriodStartDate",
    "PeriodEndDate",
    "CountValue",
    "PartOfCumulativeCountSeries",
    "Fatalities"
]
```

```python
import pandas as pd

url = (
    f"https://www.tycho.pitt.edu/api/query?"
    f"apikey={API_KEY}"
    f"&ConditionName=Measles"
    f"&CountryISO=US"
    f"&Admin1ISO=US-PA"
    f"&PeriodStartDate%3E=1960-01-01"
    f"&PeriodEndDate%3C=1970-12-31"
)

df = pd.read_csv(url)[core_columns]

print(df.shape)
df.head()
```
